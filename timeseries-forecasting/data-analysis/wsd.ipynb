```python
import pandas as pd
```

```python
df = pd.read_csv("../experiment/data/WSD.csv")
df.head()
```

```python
# only consider real-world time series
real_world = [ts for ts in df.ts.unique() if "real-world" in ts]

print(f"Total of : {len(df.ts.unique())}")
print(f"Real-world time series selected: {len(real_world)}")

real_df = df[df["ts"].isin(real_world)]

assert len(real_df.ts.unique()) == len(real_world)
```

```python
# Generate WSD version for time series forecasting task containing only the first 3000 values
wsd_reduced = real_df.sort_values(by=["ts", "dt#"]).groupby('ts').head(3000).reset_index(drop=True)
wsd_reduced.to_csv("../experiment/data/WSD_3k.csv", index=False)

assert len(wsd_reduced) == (len(real_df.ts.unique()) * 3000)
```
