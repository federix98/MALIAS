```python
for system in ["linux2_i7_4790", "linux3_e3_1240", "openbsd1_i7_4790"]:
    for log_path in Path(f"./{system}/").rglob("*.json"):
        if log_path.is_file():
            with open(log_path, "r") as in_f:
                print(log_path)
                data = json.load(in_f)
                file = log_path.stem

                for bench in data["wallclock_times"].keys():
                    timeseries = []
                    for rep in range(np.array(data["wallclock_times"][bench]).shape[0]):        
                        vals = data["wallclock_times"][bench][rep]
                        timeseries.append(vals)
                        #print(bench, rep)
                    with open(f"data/timeseries/all/{file}___{bench}.json".replace(":", "-"), "w") as out_file:
                        json.dump(timeseries, out_file)
```
