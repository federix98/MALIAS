```python
import pandas as pd
```

```python
### FLAML - Time Series Forecasting - Times

tsf_fit = pd.read_csv("./results/flaml_exp_AzFuncInvoke.csv")
tsf_refit = pd.read_csv("./results/flaml_exp_predict_AzFuncInvoke.csv")
tsf_train = pd.merge(left=tsf_fit, right=tsf_refit, on='ts', how='inner')
num_testing_windows = 102
tsf_train["full_training_time"] = tsf_train["fit_time"] + (tsf_train["refit_time"] / num_testing_windows)
avg_training_time_with_refit = tsf_train["full_training_time"].mean() / 1000 # for this task, times are in milliseconds

print("FLAML Time Series Forecasting - Times analysis")
print("---- FIT ----")
print("Fitting time mean for time series (considering re-fit)", avg_training_time_with_refit)
print("---- PREDICT ----")
print("Predict time mean for time series", tsf_refit.pred_time.mean() / 1000)
```

```python
### FLAML - Anomaly Detection
tsad_metrics = pd.read_csv("./results/flaml_tsad_metrics.csv")

print("FLAML Anomaly Detection - Times analysis")
print("---- FIT ----")
print("Fitting time mean for time series", tsad_metrics.train_time.mean())
print("---- PREDICT ----")
print("Predict time mean for time series", tsad_metrics.test_time.mean())
```

```python
### FLAML - Steady State Detection
ssd_fit = pd.read_csv("./ssd_results/models_metrics_fit.csv")
ssd_pred = pd.read_csv("./ssd_results/models_metrics_forkwise_eval.csv")

num_of_testing_series_in_fold = 531

print("FLAML Times analysis")
print("---- FIT ----")
print("Fitting time mean for each fold", ssd_fit.train_time.mean())
print("Fitting time mean for time series", ssd_fit.train_time.mean() / num_of_testing_series_in_fold)
print("---- PREDICT ----")
print("Predict time mean for the entire dataset", ssd_pred.test_time.sum())
print("Predict time mean for time series", ssd_pred.test_time.mean())
```
