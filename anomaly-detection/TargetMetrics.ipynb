```python
import numpy as np
import pandas as pd
import json
```

```python
datasets = ["AIOPS", "WSD"]
models = ["AR", "LSTMADalpha", "FCVAE", "FITS", "EncDecAD"]
target_score = "3-th auprc under event-based pa with mode squeeze"
```

```python

for dataset in datasets:
    res = []
    for model in models:
        metrics_path = f"Results/Evals/{model}/naive/{dataset}/all.json"
        with open(metrics_path) as m_file:
            metrics = json.loads(m_file.read())
        assert metrics
        for timeseries_name, timeseries_data in metrics.items():
            res.append({
                "id": f"{timeseries_name}",
                "model": model,
                "score": float(timeseries_data[target_score])
                #"score": float(timeseries_data[target_score]["f1"])
            })
    res_df = pd.DataFrame(res)
    res_df = res_df.pivot_table(values="score", index="id", columns="model")
    res_df.to_csv(f"{dataset}_metrics.csv")
```
